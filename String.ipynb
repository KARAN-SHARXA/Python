## String are the sequence of Character

```python
s = 'Hello, World!'
s = "hello, world!"
# for multiline strings, use triple quotes
s = '''hello
    world!'''
print(s)
s = str('hello')
```

# Accessing Substring from a string

```python
s = 'hello, world!'
print(s[0]) # prints the first character of the string

print(s[2:5]) # prints characters from index 2 to 4
print(s[-5:-2]) # prints characters from index -5 to -3
```

```python
# Editing and deleting strings
s = 'hello, world!'
s[0] = 'H' # this will raise an error because strings are immutable4
print(s) # prints 'hello, world!'
```

```python
s = 'Hello, World!'
del s
print(s) # this will raise an error because the string has been deleted
```

```python
# Operating On strings
# Arithmetic operations on strings
# logical operations on strings
# loops on strings

print('hello' +' '+ 'world') # prints 'helloworld'
```

```python
# Common Functions on strings
s = 'hello, world!'
print(len(s)) # prints the length of the string
print(max(s)) # prints the maximum character in the string
print(min(s)) # prints the minimum character in the string
print(sorted(s, reverse=True)) # prints the sorted characters in the string in descending order
```

```python
# Capialize, title, upper, lower, swapcase, casefold, isupper, islower, istitle, isspace, isalpha, isdigit, isalnum, isdecimal, isnumeric
s = 'Hello, World!'
print(s.capitalize()) # prints 'Hello, world!'
print(s.title()) # prints 'Hello, World!'
print(s.upper()) # prints 'HELLO, WORLD!'
print(s.lower()) # prints 'hello, world!'
print(s.swapcase()) # prints 'hELLO, wORLD!'
print(s.casefold()) # prints 'hello, world!'
```

```python
# Count / find , index
s = 'hello, world!'
print(s.count('l')) # prints the number of occurrences of 'l' in the string
print(s.find('l')) # prints the index of the first occurrence of 'l' in the string
print(s.index('l')) # prints the index of the first occurrence of 'l' in the string
```

```python
#startswith, endswith, isalnum, isalpha, isdecimal, isdigit, isidentifier, islower, isnumeric, isprintable, isspace, istitle, isupper
s = 'hello, world!'
print(s.startswith('hello')) # prints True
print(s.endswith('world!')) # prints True
print(s.isalnum()) # prints False
print(s.isalpha()) # prints False
print(s.isdecimal()) # prints False
print(s.isdigit()) # prints False
```

```python
name = 'karan'
gender = 'male'
print('hi my name is {} and i am a {}'.format(name, gender)) # prints 'hi my name is karan and i am a male'
```

```python
# Split, join, replace, strip, lstrip, rstrip
s = 'hello, world!'
print(s.split(',')) # prints ['hello', ' world!']
print(','.join(['hello', 'world!'])) # prints 'hello,world!'
print(s.replace('world', 'python')) # prints 'hello, python!'
print(s.strip()) # prints 'hello, world!'
```
