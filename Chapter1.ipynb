# DATA TYPE

<h6>Data Type in Python: A data type defines the type of value a variable can store, such as numbers, text, or Boolean values.<h6/>

```python
#INTEGER
print(8)
#DECIMAL/FLOAT
print(7.7)
#Boolean
print(True)
print(False)
#Text/string
print('hello world')
#complex
print(2+3j)
```

```python
#type() function is used to find the data type of a variable or value.
type(7)
type(7.7)
```

# Variables

<h6>Static Typing: The variable's data type is specified and checked before execution, e.g., C++.



Dynamic Typing: The variable's data type is determined automatically at runtime, e.g., Python.<h6/>

```python

name = "karan" #string
age = 20 #int
print(name)
print(age)
```

```python
a = 1
b = 2
c = 3
print( a , b , c )
a , b , c = 4 , 5 , 6
print( a , b , c)
a=b=c= 5
print( a , b , c)
```

# Keywords and indentifier

<h6>Keyword: A reserved word in Python that has a special meaning, e.g., if, else, for.



Identifier: A name used to identify a variable, function, or class, e.g., age, name.<h6/>

<h6>Type Conversion in Python: Type conversion is the process of changing a value from one data type to another.

Implicit Conversion: Python automatically converts the data type.



Explicit Conversion: The programmer manually converts the data type using functions like int(), float(), and str()<h6/>

```python
#input
a=int(input("Press Enter to exit..."))
b=int(input("Press Enter to exit..."))
print(a+b)


# type Conversion
#Implicit Type Conversion and Explicit Type Conversion
print(5+5.6)
print(4+int(4))
print(type(a))
```
