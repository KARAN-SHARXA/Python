## Operators in Python

<h6>Operator in Python: An operator is a symbol or keyword used to perform operations on values and variables.<h6/>
<h7>1. Arithmetic Operators: Used to perform mathematical calculations like addition, subtraction, multiplication, and division.

2. Relational Operators: Used to compare two values.

3. Logical Operators: Used to combine or reverse conditions.

4. Bitwise Operators: Used to perform operations on the binary bits (0s and 1s) of numbers.

5. Assignment Operators: Used to assign values to variables.

6. Membership Operators: Used to check whether a value exists in a sequence or collection.<h7/>

```python
print(5+5)
print(5+5.0)
print(5+int(5.0))
print(6-6)
print(5/2)
print(5//2)
print(5%2)
print(5**2)
```

```python
# Relation Operators
print(5>6)
print(5<6)
print(5>=6)
print(5<=6)
print(5==6)
print(5!=6)
```

```python
# Logical Operators
print(1 and 0)
print(1 or 0)
print(not 1)
```

```python
# Bitwise Operators
# bitwise and (&)
print(5&6)

# bitwise or (|)
print(5|6)

# bitwise xor (^)
print(5^6)

# bitwise not (~)
print(~5)

# bitwise left shift (<<)
print(5<<1)


# bitwise right shift (>>)
print(5>>1)
```

```python
# assignment Operators
a=5

a+=5
print(a)

# a++ and ++a is not supported in python. We can use a+=1 instead of a++ and a-=1 instead of a--

```

```python
# Membership Operators
# in

print('D' not in 'karan')

print('D' in 'karan')
```
