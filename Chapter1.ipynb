# Variables

```python
# static vs Dynamic typing
# In static typing, the data type of a variable is known at compile time. 
name = "karan" #string
age = 20 #int
print(name)
print(age)
```

```python
# Dynamic Typing
a = 5
```

```python
# dynamic Binding
# In dynamic typing, the data type of a variable is determined at runtime.  
# static typing is used in languages like C, C++, and Java, while dynamic typing is used in languages like Python, JavaScript, and Ruby.
a = 5
print(a)
b = "karan"
print(b)
```

```python
a = 1
b = 2
c = 3
print( a , b , c )
a , b , c = 4 , 5 , 6
print( a , b , c)
a=b=c= 5
print( a , b , c)
```
