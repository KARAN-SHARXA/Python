2. Relational Operators: Used to compare two values.

3. Logical Operators: Used to combine or reverse conditions.

4. Bitwise Operators: Used to perform operations on the binary bits (0s and 1s) of numbers.

<h6>A for loop is used to iterate over a sequence or iterable, whereas a while loop executes a block of code repeatedly as long as a specified condition remains True.

Remember: Use a for loop to iterate over items and a while loop to repeat based on a condition.<h6/>

<h6>while loop = It is used to execute a block of statements repeatedly until a given condition is satisfied<h6/>

```python
count = 0
while count < 5:
  print(count)
  count = count + 1
```

```python
count = 0
while(count<5):
  print(count)
  count = count + 1
else:
  print("count is greater than 5")
```

<h6>Break statement = We use break when we like to get out of or stop the loop<h6/>

```python
count = 0
while count < 5:
  print(count)
  count = count + 1
  if count == 3:
    break
else:
  print("count is greater than 5")
```

<h6>Continue: With the continue statement we can skip the current iteration, and continue with the next<h6/>

```python
count = 0
while count < 5:
    if count == 3:
        count = count + 1
        continue

    print(count)
    count = count + 1
```

```python
# extend -> adding multiple items to the end of the list
l = [1,2,3,4,5]
l.extend([6,7,8])
print(l) # [1, 2, 3, 4, 5, 6, 7, 8]
```

```python
# insert

l = [1,2,3,4,5]
l.insert(2, 2.5) # inserting 2.5 at index 2
print(l) # [1, 2, 2.5, 3, 4, 5]
```

```python
# editing an item in the list
l = [1, 2, 3, 4, 5]
l[2] = 3.5 # editing the item at index 2
print(l) # [1, 2, 3.5, 4, 5]
```

```python
# deleting an item from the list
# del statement removes an item at a specified index
l = [1, 2, 3, 4, 5]
del l[2] # deleting the item at index 2
print(l) # [1, 2, 4, 5]
del l[1:3] # deleting items from index 1 to index 2
print(l) # [1, 5]


#remove() method removes an item from the list by value
l = [1, 2, 3, 4, 5]
l.remove(3) # removing 3 from the list
print(l) # [1, 2, 4, 5]

# pop() method removes an item at the specified index and returns it
l = [1, 2, 3, 4, 5]
l.pop(2) # removing the item at index 2
l.pop() # removing the last item
print(l) # [1, 2, 4]
l= [1, 2, 3, 4, 5]
l.clear()
print(l) # []
```

```python
# Operators in List
# Arithmetic Operators
# memnership operators

l1 = [1, 2, 3]
l2 = [4, 5, 6]
l3 = l1 + l2 # concatenation
l4 = l1 * 2 # repetition
print(l3) # [1, 2, 3, 4, 5, 6]
print(l4) # [1, 2, 3, 1, 2, 3]
```

```python
# membership operators
l1 = [1, 2, 3]
l2 = [4, 5, 6]
print(2 in l1) # True
print(2 in l2) # False
print(4 not in l1) # True
print(4 not in l2) # False
```

```python
# list functions

l = [3, 1, 4, 1, 5]
print(len(l)) # 5
print(max(l)) # 5
print(min(l)) # 1
print(sum(l)) # 14
print(l.count(1)) # 2
print(l.index(4)) # 2
print(l.sort()) # [1, 1, 3, 4, 5]
print(l.reverse()) # [5, 4, 3, 1, 1]
print(sorted(l, reverse=True)) # [5, 4, 3, 1, 1]
print(l.copy()) # [5, 4, 3, 1, 1]
```
